```python
import numpy as np
import matplotlib.pyplot as plt

# Euler Method
def euler(f, t0, y0, h, t_end):
    t = np.arange(t0, t_end + h, h)
    y = np.zeros(len(t))

    y[0] = y0

    for i in range(len(t) - 1):
        y[i + 1] = y[i] + h * f(t[i], y[i])

    return t, y


# Heun Method (Improved Euler)
def heun(f, t0, y0, h, t_end):
    t = np.arange(t0, t_end + h, h)
    y = np.zeros(len(t))

    y[0] = y0

    for i in range(len(t) - 1):
        k1 = f(t[i], y[i])
        y_predict = y[i] + h * k1
        k2 = f(t[i] + h, y_predict)
        y[i + 1] = y[i] + (h / 2) * (k1 + k2)

    return t, y


# Midpoint Method
def midpoint(f, t0, y0, h, t_end):
    t = np.arange(t0, t_end + h, h)
    y = np.zeros(len(t))

    y[0] = y0

    for i in range(len(t) - 1):

        # Initial slope
        k1 = f(t[i], y[i])

        # Estimate the midpoint
        y_mid = y[i] + (h / 2) * k1
        t_mid = t[i] + (h / 2)

        # Slope at the midpoint
        k2 = f(t_mid, y_mid)

        # Update solution
        y[i + 1] = y[i] + h * k2

    return t, y


# Print Result Table
def print_table(title, t, numerical, exact_function, max_rows=5):

    print("\n" + "="*70)
    print(title)
    print("="*70)

    print(f"{'t':>8} {'Numerical':>15} {'Exact':>15} {'Abs Error':>15}")

    # Calculate how many indices to skip
    skip = max(1, len(t) // max_rows)

    for i in range(0, len(t), skip):

        exact = exact_function(t[i])
        error = abs(exact - numerical[i])

        print(f"{t[i]:8.3f} {numerical[i]:15.6f} {exact:15.6f} {error:15.6f}")

    # Ensure the last point is always shown
    if (len(t) - 1) % skip != 0:
        i = len(t) - 1
        exact = exact_function(t[i])
        error = abs(exact - numerical[i])

        print(f"{t[i]:8.3f} {numerical[i]:15.6f} {exact:15.6f} {error:15.6f}")


# Plot Results
def plot_results(t_exact,
                 y_exact,
                 t_euler,
                 y_euler,
                 t_heun,
                 y_heun,
                 t_mid,
                 y_mid,
                 title):

    plt.figure(figsize=(10,6))

    plt.plot(t_exact,
             y_exact,
             'k',
             linewidth=3,
             label="Exact")

    plt.plot(t_euler,
             y_euler,
             'ro--',
             label="Euler")

    plt.plot(t_heun,
             y_heun,
             'bs--',
             label="Heun")

    plt.plot(t_mid,
             y_mid,
             'g^-',
             label="Midpoint")

    plt.xlabel("t")
    plt.ylabel("y")
    plt.title(title)

    plt.grid(True)
    plt.legend()

    plt.show()
```

```python
# Differential Equation
def f(t, y):
    return (1 + 4*t) - y

# Exact Solution
def exact(t):
    return 4*t - 3 + 4*np.exp(-t)

# Parameters
t0 = 0
y0 = 1
t_end = 5

# Step sizes required by the lab
step_sizes = [1, 0.5, 0.25, 0.1, 0.01, 0.001]

for h in step_sizes:

    print("\n" + "="*80)
    print(f"RESULTS FOR STEP SIZE h = {h}")
    print("="*80)

    # Solve using all three methods
    t_euler, y_euler = euler(f, t0, y0, h, t_end)
    t_heun, y_heun = heun(f, t0, y0, h, t_end)
    t_mid, y_mid = midpoint(f, t0, y0, h, t_end)

    # Print result tables
    print_table("Euler Method", t_euler, y_euler, exact)
    print_table("Heun Method", t_heun, y_heun, exact)
    print_table("Midpoint Method", t_mid, y_mid, exact)

    # Exact solution for smooth curve
    t_exact = np.linspace(t0, t_end, 500)
    y_exact = exact(t_exact)

    # Plot comparison
    plot_results(
        t_exact,
        y_exact,
        t_euler,
        y_euler,
        t_heun,
        y_heun,
        t_mid,
        y_mid,
        f"Lab Task 1 (h = {h})"
    )
```
